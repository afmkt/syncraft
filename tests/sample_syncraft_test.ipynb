# Visualizing Syncraft Syntax and AST
This quick notebook shows how to use the helper utilities in `syncraft/dev.py` to inspect grammar definitions (`Syntax`) and the abstract syntax trees (AST) they produce. The helpers work both in plain text and as SVG diagrams for richer visualisation.

## Setup
Import the helpers and choose a sample grammar. Feel free to swap in your own `Syntax` definitions.

```python
import sys
from pathlib import Path
try:
    base_path = Path(__file__).resolve()
except NameError:
    base_path = Path.cwd().resolve()
if base_path.is_file():
    search_root = base_path.parent
else:
    search_root = base_path
for candidate in [search_root, *search_root.parents]:
    if (candidate / "pyproject.toml").exists():
        PROJECT_ROOT = candidate
        break
else:
    PROJECT_ROOT = search_root
if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))
PROJECT_ROOT
```

```python
from syncraft.syntax import Syntax
from syncraft.dev import rich_parser, syntax2svg, ast2svg
from syncraft.sqlite3 import expr  # replace with your own grammar if desired
```

## Plain-text view of the Syntax tree
`rich_parser` uses the new `Syntax.spec` metadata to pretty-print the structure via simple indentation.

```python
# NBVAL_IGNORE_OUTPUT
rich_parser(expr, max_lines=80, max_depth=8, use_rich=False)
```

## SVG railroad diagram
If the optional `railroad-diagrams` package is installed, `syntax2svg` returns an SVG snippet that we can display inline.

```python
# NBVAL_IGNORE_OUTPUT
from IPython.display import SVG, display
try:
    svg_data = syntax2svg(expr, max_nodes=400, max_depth=10)
except Exception as exc:
    svg_data = None
    print(f"SVG rendering unavailable: {exc}")
if svg_data:
    display(SVG(svg_data))
else:
    print("Install `railroad-diagrams` to enable SVG rendering.")
```

## Parse input to inspect the AST
Use the parser to obtain an AST value. We'll reuse the SQLite expression grammar for demonstration.

```python
from syncraft.adapter import sqlglot_lex, sqlglot_token_class
from syncraft.parser import parse

sql = "1 + 2 * 3"
tokens = sqlglot_lex(sql)
token_cls = sqlglot_token_class()
ast_value, _ = parse(expr, tokens, token_class=token_cls)
ast_value
```

## Visualize the AST as SVG
`ast2svg` converts the AST into a Graphviz diagram when the `graphviz` package and binaries are available.

```python
# NBVAL_IGNORE_OUTPUT
from IPython.display import SVG, display

ast_svg = ast2svg(ast_value)
if ast_svg:
    display(SVG(ast_svg))
else:
    print("Install `graphviz` (package and system binary) to enable AST SVG rendering.")
```
