```python
# 🌳 Testing AST Integration with _repr_html_
import sys
import importlib

# Reload AST module to get the new _repr_html_ method
if 'syncraft.ast' in sys.modules:
    del sys.modules['syncraft.ast']

from syncraft.regex import parse
from syncraft.dev import ast2svg

print("=== Testing AST automatic display ===")

# Parse a simple regex to get an AST
ast_result = parse('a+', raw=True)
print("AST type:", type(ast_result))
print("Has _repr_html_:", hasattr(ast_result, '_repr_html_'))

print("\n🌳 AST Tree Diagram:")
print("This should automatically show a beautiful tree visualization!")

# This should now display as a styled AST tree diagram
ast_result
```

```python
# 🌳 Complete AST Module Reload and Test
import sys

# Remove AST and related modules from cache
ast_modules = [mod for mod in sys.modules.keys() if 'ast' in mod or mod.startswith('syncraft')]
print("Reloading modules:", len(ast_modules))

for module in ast_modules:
    if module in sys.modules:
        del sys.modules[module]

# Fresh imports
from syncraft.regex import parse
print("✅ Modules reloaded")

# Test AST object
ast_result = parse('ab*', raw=True)
print("AST has _repr_html_:", hasattr(ast_result, '_repr_html_'))

# Test manual call
if hasattr(ast_result, '_repr_html_'):
    html_content = ast_result._repr_html_()
    print("Manual _repr_html_ works:", html_content is not None and len(html_content) > 1000)

print("\n🌳 AST automatic display:")
ast_result
```
