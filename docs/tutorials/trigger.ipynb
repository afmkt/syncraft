# Syncraft Quickstart Tutorial

Welcome to Syncraft! This tutorial will show you how to define a grammar, parse input, and visualize results.

```python
from syncraft import literal, choice, generate, parse_word
from syncraft.dev import syntax2svg, ast2svg
from IPython.display import SVG, display
```

## Define a Grammar

```python
A = literal('a')
B = literal('b')
syntax = A + B  # Sequence
```

## Visualize the Grammar

```python
svg = syntax2svg(syntax)
display(SVG(svg)) if svg else print('Railroad-diagrams not installed.')
```

## Parse Input and Visualize AST

```python
ast, _ = parse_word(syntax, 'a b')
svg_ast = ast2svg(ast)
display(SVG(svg_ast)) if svg_ast else print('Graphviz not installed.')
```
